# SyllabusQA Benchmark Results
Comparison table of model performance across different execution modes

```python
import pandas as pd
import json
from pathlib import Path

# Load results
results_path = Path('notebook-lm-clone/outputs/syllabusqa_results/results_20260429_234944.json')

with open(results_path) as f:
    results = json.load(f)

# Create summary dataframe
data = []
for result in results:
    data.append({
        'Model': result['model_name'],
        'Mode': result['mode'].replace('_', '-').upper(),
        'Ground Truth': result['ground_truth'][:30] + '...',
        'Output Generated': 'Yes' if result.get('prediction') else 'No',
        'Output Length': len(result.get('prediction', ''))
    })

df = pd.DataFrame(data)

# Display the table
print("="*120)
print(f"{'Model':<25} | {'Mode':<12} | {'Ground Truth':<35} | {'Output':>8} | {'Length':>8}")
print("="*120)

for _, row in df.iterrows():
    print(f"{row['Model']:<25} | {row['Mode']:<12} | {row['Ground Truth']:<35} | {row['Output Generated']:>8} | {row['Output Length']:>8}")

print("="*120)
```
